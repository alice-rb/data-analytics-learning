#### Ejercicio 1: Calculadora de índice de masa corporal
- Función que calcule el IMC escrito por el usuario.
- La función ha de clasificar el resultado en sus respectivas categorias:
        Bajo peso: menos de 18,5.
        Peso normal: 18,5-24,9.
        Sobrepeso: 25-29,9.
        Obesidad: más de 30.
- Cálculo del IMC: kg / (altura**2)

```python
def datos_usuario():
    try:
        peso = float(input( 'Por favor, indica tu peso en kg: '))
        altura = float(input('Por favor, indica tu altura en metros: '))
        return peso,altura
    except ValueError:
        print("Tipo de dato introducido erróneo")

def calculadora_IMC(peso,altura):
    '''
    Calcula el IMC en base al peso y altura introducidos
    peso float,int
    altura float,int
    '''
    IMC = round(peso / (altura**2),2)
    return IMC

def clasificador_IMC(IMC):
    if IMC < 18.5:
        print(f'Clasificación: Bajo peso {IMC}')
    elif 18.5 <= IMC <= 24.9:
        print(f'Clasificación: Peso normal {IMC}')
    elif 25 <= IMC <= 29.9:
        print(f'Clasificación: Sobrepeso {IMC}')
    else:
        print(f'Clasificación: Obesidad {IMC}')

peso_usuario,altura_usuario = datos_usuario()
IMC_usuario = calculadora_IMC(peso_usuario,altura_usuario)
clasificador_IMC(IMC_usuario)
```

#### Ejercicio 2: Convertidor de temperaturas
- Selecciona almenos 2 conversores de temperatura, de tal manera que al introducir una temperatura devuelva, como mínimo, dos conversiones que se puedan almacenar. 
	- Existen diversas unidades de temperatura. Las más comunes son: Celsius (°C), Fahrenheit (°F) y Kelvin (K). También existen otras unidades como Rankine (°Ra) i Réaumur (°Re).
- (EXTRA): Piensa una manera de almacenar todas las posibles conversiones en un solo objeto.

```python
def registrar_datos(diccionario,**datos):
    '''
    Función para almacenar datos en un diccionario
    '''
    for clave, valor in datos.items():
        diccionario.setdefault(clave, []).append(valor)

# Funciones de conversión de temperaturas + función de registro de datos para ir almacenando los resultados
def conversor_celsius(temperatura,diccionario):
    fahrenheit = round((temperatura * (9/5)) + 32,2)
    kelvin = round(temperatura + 273.15,2)
    registrar_datos(diccionario,celsius=temperatura,fahrenheit=fahrenheit,kelvin=kelvin)
    return fahrenheit,kelvin

def conversor_farenheit(temperatura,diccionario):
    celsius = round((temperatura - 32) * (5/9),2)
    kelvin = round(celsius + 273.15,2)
    registrar_datos(diccionario,celsius=celsius,fahrenheit=temperatura,kelvin=kelvin)
    return celsius,kelvin

def conversor_kelvin(temperatura,diccionario):
    celsius = round(temperatura - 273.15,2)
    fahrenheit = round((celsius * (9/5)) + 32,2)
    registrar_datos(diccionario,celsius=celsius,fahrenheit=fahrenheit,kelvin=temperatura)
    return celsius,fahrenheit

def selector_temperatura(diccionario):
    '''
    Función interactiva para establecer qué unidad de temperatura serán los grados que se solicitarán posteriormente.\n
    Si se elige una opción fuera de 1,2,3, seguirá solicitando una opción válida.\n
    Dependiendo de la opción elegida, se ejecutará una función conversor de unidades u otra.
    '''
    while True:
        print("\n ¿Qué unidad de temperatura utilizamos?")
        print("1. Celsius")
        print("2. Farenheit")
        print("3. Kelvin")

        opcion = input("Selecciona una unidad de temperatura: ")
        
        if opcion in ("1","2","3"):
            break
        else:
            print("Opción  no válida. Por favor, seleccione de nuevo")
    try:
        grados = float(input("¿Qué temperatura hace?"))
    except ValueError:
        print("Temperatura introducida no válida")
        return

    if opcion == "1":
        conversor_celsius(grados,diccionario)
    elif opcion == "2":
        conversor_farenheit(grados,diccionario)
    elif opcion == "3":
        conversor_kelvin(grados,diccionario)

    return diccionario

# Creamos un diccionario vacío anexamos las temperaturas según el tipo de unidad
temperaturas_totales = {}

# Llamamos a la función de selección, donde elegiremos qué unidad transformar 
selector_temperatura(temperaturas_totales)

```

#### Ejercicio 3: Contador de palabras de un texto
- Escribe una función que dado un texto, muestre las veces que aparece cada palabra. Intenta que se gestionen todas las problemáticas posibles que hagan que el programa no funcione correctamente.
- (EXTRA): Cual es el largo de media de las palabras del texto que has escrito?

```python
def limpiar_palabras(texto_lista):
    '''
    Función para asegurar que únicamente se tengan en cuenta letras del alfabeto \n
    str, list
    '''
    # -- Esta parte del código está pensada por la reutilización de funciones en otros ejercicios
    if isinstance(texto_lista, list):
        texto = " ".join(texto_lista)
    else:
        texto = texto_lista
    # -----------------------------------
    palabras_limpias = []

    for palabra in texto.lower().split():
        # Recorre cada palabra del texto y revisa que no tenga números ni símbolos ni al inicio ni al final para dejarla únicamente con letras
        palabra_limpia = re.sub(r'^[^a-záéíóúüñ]+|[^a-záéíóúüñ]+$', '', palabra)
        # Para evitar que se almacenen palabras vacías y luego el contador/media den valores no válidos
        if palabra_limpia:  
            palabras_limpias.append(palabra_limpia)
    return palabras_limpias

def contador_valores(dic,clave,valor):
    ''' 
    Función para acumular valores en un diccionario\n
    dict
    '''
    dic[clave] = dic.get(clave,0) + valor
    return dic

def contador_palabras(texto):
    contador = {}
    for palabra in limpiar_palabras(texto):
        contador_valores(contador,palabra,1)
    return contador

def media_palabras(texto):
    texto_limpio = limpiar_palabras(texto)
    letras = sum(len(palabra) for palabra in texto_limpio)
    media = round(letras / len(texto_limpio))
    return media

frase = input("Escribe una frase: ")
print(f"La frase \"{frase}\" contiene: {contador_palabras(frase)}, y la media del largo de las palabras es de {media_palabras(frase)}")
```

#### Ejercicio 4: Diccionario inverso (con posibilidad de duplicados)

- Intercambio de claves-valor. Los valores y claves en el diccionario original son únicos; si este no fuera el caso, la función debería imprimir un mensaje de advertencia, junto con una lista con los valores asociados a la clave repetida.

```python
def diccionario_inverso(diccionario):
    diccionario_nuevo = {}
    # Recorremos los items del diccionario pasado como argumento y, si la nueva clave ya tiene un valor existente, lo anexamos y mostramos un mensaje de advertencia
    for clave,valor in diccionario.items():
        if valor in diccionario_nuevo:
            diccionario_nuevo[valor].append(clave)
            print(f"Cuidado! La clave {valor} tiene asociados: {diccionario_nuevo[valor]} ")
        else:
            diccionario_nuevo[valor] = [clave]
    return diccionario_nuevo

encuesta = {"Mario": 4, "Alejandra": 7, "Alex": 9, "Nuria": 5, "Pedro": 4, "Sofia": 9}
diccionario_inverso(encuesta)

```

#### Ejercicio 2: Conversión de tipos de datos

- A partir de una lista de datos, necessitamos generar dos listas:
    - Una donde estén todos los elementos que se pueden convertir en flotantes.
    - Otra con los elementos que no ha sido posible convertir.

```python
def convertir_flotante(listado):
    lista_flotantes = []
    lista_no_flotantes = []
    # Primero verificamos que tipo de dato hay en el listado pasado como argumento
    for elemento in listado:
        # En caso de haber elementos anidados dentro del listado, hacemos uso de la recursividad, donde se vuelve a llamar a la función para que los desglose
        if isinstance(elemento,(list,tuple)): 
            flot, not_flot = convertir_flotante(elemento)
            lista_flotantes.extend(flot) # -- usamos extend para anexar los elementos de los objetos actuales a la lista que estamos construyendo de manera individual
            lista_no_flotantes.extend(not_flot)
            continue # -- para que el siguiente bloque de codigo se ejecute cuando los elementos anidados ya se hayan procesado 

        try:
            lista_flotantes.append(float(elemento))
        except ValueError or TypeError: # -- los valores imposible de convertir en flotante por su tipo
            lista_no_flotantes.append(elemento)
    
    return lista_flotantes,lista_no_flotantes

lista_recibida = ['1.3','one','1e10','seven','3-1/2',('2',1,1.4,'not-a-number'),[1,2,'3','3.4']]
decim, no_decim = convertir_flotante(lista_recibida) 
print(f"Valores convertibles: {decim}\nValores no convertibles: {no_decim}")


```

#### Ejercicio 1: Generador de contraseñas

- Crea una función que genere contrasenyes seguras. La función ha de depender de los siguientes parámetros:

    - longitud (int): Longitud 
    - mayúsculas (bool = True): Si han de aparecer mayúsculas
    - minúsculas (bool = True): Si han de aparecer minúsculas
    - números (bool = True): Si han de aparecer números
    - signos (bool = False): Si han de aparecer carácteres especiales (,-$? o similares)

- (EXTRA) Que la función copie la contraseña automáticamente al portapapeles del ordenador

```python
def crear_clave(longitud=10,mayus=True,minus=True,nums=True,signos=False):
    # Como si crearamos una lista con todos los elementos que pueden estar incluídos en la contraseña (en caso de ser True)
    caracteres = "".join([
        string.ascii_uppercase if mayus else "",
        string.ascii_lowercase if minus else "",
        string.digits if nums else "",
        string.punctuation if signos else ""
    ])

    # La clave se creará a partir de la iteración (por la longitud especificada) de manera aleatoria en nuestra lista con todos los elementos posibles para la contraseña
    clave_generada = "".join(random.choice(caracteres) for _ in range(longitud))

    return clave_generada

mi_clave = crear_clave(10,True,True,True,False)
pyperclip.copy(mi_clave)
print(f"Tu clave ha sido generada: {mi_clave}. Recuerda que la tienes copiada en el portapapeles (ctrl+V)")

```

#### Ejercicio 2: Procesamiento de datos simple

- A partir del histórico de partidos de fútbol en un arhivo, necessitamos procesar los datos de manera automática para extraer los resultados.

    - Utiliza el archivo "historic_partits.txt"
    - El programa ha de devolver:
        1. Número total de goles de cada equipo.
        2. Nombre del equipo con más goles marcados
        3. Nombre del equipo con más goles encajados
        4. Clasificación global (victoria: 3 pts, empate: 1 pts, derrota: 0 pts)
        
Aviso: En este código se reutiliza la función *contador_valores* del ejercicio 1.3 y *leer_archivo* del ejercicio 2.1

```python
def ordenar_partidos(archivo):
    partidos = {}
    # Recorremos cada linea del archivo que hemos pasado como argumento, enumerando los partidos empezano por el 1
    for num_partido,partido in enumerate(archivo,1):
        # Limpiamos las lineas de cosas que dificulten su almacenaje y luego dividimos la información para almacenarla de manera organizada
        partido = partido.strip()
        equipo1, marcador, equipo2 = partido.split('\t') 
        goles1, goles2 = map(int, marcador.split('-'))
        # Resultado: un diccionario con número de partido y dentro, otro diccionario con cada equipo participante y los goles marcados
        partidos[f"P_{num_partido}"] = {equipo1 : goles1,
                                        equipo2 : goles2}
    return partidos

def equipo_goles(archivo):
    partidos = ordenar_partidos(archivo)
    # Creamos un diccionario para registrar tanto los goles marcados como los encajados para cada equipo a partir de los partidos
    goles_marcados = {}
    goles_encajados = {}

    # Iteramos dentro de los valores del diccionario con los partidos (diccionario anidado de equipo + goles = puntuacion), para asignar los goles según marcados o encajados para cada equipo
    for puntuacion in partidos.values(): 
        (equipo1, goles1), (equipo2, goles2) = puntuacion.items()
        contador_valores(goles_marcados,equipo1,goles1)
        contador_valores(goles_marcados,equipo2,goles2)
        contador_valores(goles_encajados,equipo1,goles2)
        contador_valores(goles_encajados,equipo2,goles1)

    return goles_marcados, goles_encajados

def equipo_clasificacion(archivo):
    # Obtenemos el diccionario con los partidos ordenados y creamos uno nuevo vacío para incorporar los puntos para cada equipo
    partidos = ordenar_partidos(archivo)
    puntos = {}

    # Iteramos dentro de los valores del diccionario con los partidos (diccionario anidado de equipo + goles = puntuacion), para extraer equipos y goles
    for puntuacion in partidos.values(): 
        (equipo1, goles1), (equipo2, goles2) = puntuacion.items()

        if goles1 == goles2: 
            contador_valores(puntos,equipo1,1)
            contador_valores(puntos,equipo2,1)
        else:
            if goles1 > goles2:
                ganador = equipo1
            else:
                ganador = equipo2
            contador_valores(puntos, ganador, 3)

    # Ordenamos la salida del listado de más a menos puntos, asginando que se ordenen por valor descendente
    return puntos
  
resultados = leer_archivo("historic partits.txt")

goles_marcados, goles_encajados = equipo_goles(resultados)
max_goles_marcados = max(goles_marcados.items(), key=lambda resultado: resultado[1])
max_goles_encajados = max(goles_encajados.items(), key=lambda resultado: resultado[1])

df_marcados = pd.DataFrame.from_dict(goles_marcados,orient='index',columns=['Goles Marcados'])
df_encajados = pd.DataFrame.from_dict(goles_encajados,orient='index',columns=['Goles Encajados'])
df_puntos = pd.DataFrame.from_dict(equipo_clasificacion(resultados),orient='index',columns=['Puntos'])
df_resultados = df_marcados.join(df_encajados).join(df_puntos)

print(f'''El equipo con más goles registrados es: {max_goles_marcados}
El equipo con más goles encajados es: {max_goles_encajados}''')
df_resultados.sort_values('Puntos',ascending=False)

```
